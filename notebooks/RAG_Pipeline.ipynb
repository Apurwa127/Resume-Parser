```python
occupation_docs = (
                    onet.groupby(['O*NET-SOC Code', 'Title'])
                   .agg({
                       'Element Name': lambda x: list(x.unique()),
                       'Workplace Example': lambda x: list(x.unique()),
                       'Hot Technology': lambda x: list(x.unique()),
                       'In Demand': lambda x: list(x.unique())
                   })).reset_index()
```

```python
occupation_docs
```

```python
def create_document(row):
    skills = row["Element Name"]
    workplace_examples = row["Workplace Example"]
    hot_technology = row["Hot Technology"]
    in_demand = row["In Demand"]

    document = f"""
Occupation: {row["Title"]}
O*NET-SOC Code: {row["O*NET-SOC Code"]}

Software Skills:
{skills}

Workplace Examples:
{workplace_examples}

Hot Technology:
{hot_technology}

In Demand:
{in_demand}
"""

    return document.strip()


occupation_docs["document"] = occupation_docs.apply(
    create_document,
    axis=1
)

print("Number of documents:", len(occupation_docs))
print(occupation_docs["document"].iloc[0])
```
