```python
occupation_docs = (
                    onet.groupby(['O*NET-SOC Code', 'Title'])
                   .agg({
                       'Element Name': lambda x: list(x.unique()),
                       'Workplace Example': lambda x: list(x.unique()),
                       'Hot Technology': lambda x: list(x.unique()),
                       'In Demand': lambda x: list(x.unique())
                   })).reset_index()
```

```python
def create_document(row):
    skills = row["Element Name"]
    workplace_examples = row["Workplace Example"]
    hot_technology = row["Hot Technology"]
    in_demand = row["In Demand"]

    document = f"""
Occupation: {row["Title"]}
O*NET-SOC Code: {row["O*NET-SOC Code"]}

Software Skills:
{skills}

Workplace Examples:
{workplace_examples}

Hot Technology:
{hot_technology}

In Demand:
{in_demand}
"""

    return document.strip()


occupation_docs["document"] = occupation_docs.apply(
    create_document,
    axis=1
)

print("Number of documents:", len(occupation_docs))
print(occupation_docs["document"].iloc[0])
```

---------Embedding------------

```python
from sentence_transformers import SentenceTransformer
```

```python
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentence = occupation_docs['document'].tolist()
embedding_doc = embedding_model.encode(sentence)
print("embedding shape", embedding_doc.shape)
print("length", len(embedding_doc))
```

----Cosine Similarities-----------

```python
similarities = embedding_model.similarity(embedding_doc, embedding_doc)

print(similarities)
```

```python
import faiss
import numpy as np

# convert embedding to float
embedding_matrix = np.array(embedding_doc).astype(np.float32)

# Create Dimension because we have to know how many dimensions the embedding has
dimension = embedding_matrix.shape[1]
print(dimension)

# Create FAISS index
index = faiss.IndexFlatL2(dimension)

print(index)

# Add embeddings to the index: Its basically telling Faiss to store all my document embedding so that I can perform fast similarity search later
index.add(embedding_matrix)
print("Number of vectors in the index:", index.ntotal)
```

```python
def search_occupation(query, k=5):

    
    # User's questions invoke

    query_embedding = embedding_model.encode(query)

    # convert into float32
    query_embedding = np.array(query_embedding).astype("float32")

    # Make sure the shape is (1, 384)
    query_embedding = query_embedding.reshape(1, -1)

    # Perform similarity search using FAISS index

    distance, indices = faiss.IndexFlatL2.search(query_embedding, k)

    # Get the matching occupations

    results = occupation_docs.iloc[indices[0]].copy()

    results['distance'] = distance[0]

    return results
```

```python
query = "what occupations require Python?"

results = search_occupation(query, k=5)

print(results[["Title", "distance"]])
```

-------Build RAG Function------------

```python
def rag_function(query, k= 5):
    #retrieve the relevent O*Net documents based on the query
    results = search_occupation(query, k=k)

    #combine the retrieved documents
    combined_doc = " ".join(results['document'].tolist())

    #create a prompt 
    prompt = f"""
You are an AI career intelligence assistant.

Answer the user's question using the O*NET information provided below.

If the context does not contain enough information to answer the
question, say that the available context is insufficient.

O*NET Context:
{combined_doc}

User Question:
{query}
"""

    #send the context to LLM
    response = client.responses.create(
        model = "gpt-5.5",
        instructions = "You are a AI career intelligence assistant.Help the user answer their question based on the O*NET context.",
        input = prompt

    )
    return response.output_text
```

```python
answer = rag_function("What are the skills needed for data scientist?")
print(answer)
```
